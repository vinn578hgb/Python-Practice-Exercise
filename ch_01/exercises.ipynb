## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random

random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(salaries)
```

## Exercise 5: Calculating statistics and verifying
### mean

```python
mean_salary = sum(salaries) / len(salaries)
print (mean_salary)
```

```python
import statistics
```

```python
from statistics import mean

```

```python
print (mean(salaries))
```

### median

```python
median_salaries = statistics.median(salaries)
print(median_salaries)
```

### mode

```python
from collections import Counter

salary_counts = Counter(salaries)

if salary_counts:
    max_frequency = max(salary_counts.values())
    mode_salaries = [salary for salary, count in salary_counts.items() if count == max_frequency]
    print(f"Mode(s) of salaries: {mode_salaries}")
else:
    print("The list of salaries is empty, cannot calculate mode.")
```

```python
mode_salaries = statistics.mode(salaries)
print(mode_salaries)
```
